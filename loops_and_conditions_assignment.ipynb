## 🌡️ 50. Weekly Temperature Report

### Goal:
    A weather station wants to know how many days were warmer than average.

### Steps:

    Create temps = [30, 32, 28, 35, 31, 29, 36].
    
    First loop: add all the temperatures and calculate the average.
    
    Second loop: count the days above the average.
    
    Print the average (2 decimals) and the count.

### Example:

    Average = 31.57
    Days above average = 3

```python
print("\n\tWeekly Temperature Report")

temps = [30, 32, 28, 35, 31, 29, 36]
total = 0

for temp in temps:
    total += temp

average = total / len(temps)
count = 0

for temp in temps:
    if temp > average:
        count += 1

print(f"Average = {average:.2f}")
print(f"Days above average = {count}")
```
